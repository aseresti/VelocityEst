```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
v_TAFE = np.array([32.649, 18.314, 16.441, 15.389, 19.151])
v_bolus = np.array([63.040, 23.54, 58.968, 24.638, 78.993])
```

```python
np.mean(v_TAFE)
```

```python
np.mean(v_bolus)
```

```python
np.std(v_bolus)
```

```python
np.std(v_TAFE)
```

```python
area = np.pi*(0.888)**2
v_TAFE * area * 60
```

```python
v_bolus * area * 60
```

```python
mean = np.mean([v_TAFE, v_bolus], axis=0)
diff = v_TAFE - v_bolus
md = np.mean(diff)
sd = np.std(diff)

plt.rcParams['font.family'] = 'Times New Roman'
plt.rcParams['mathtext.fontset'] = 'custom'
plt.rcParams['mathtext.rm'] = 'Times New Roman'
plt.rcParams["mathtext.fontset"] = "stix"

plt.figure(figsize=(8, 6))
plt.scatter(mean, diff, s=120, color='black')

loa_upper = md + 1.96*sd
loa_lower = md - 1.96*sd
plt.text(20, loa_upper + 0.2, f'+1.96 σ = {loa_upper:.2f} (cm/s)', color='black', va='bottom', fontsize = 25)
plt.text(20, loa_lower - 0.2, f'-1.96 σ = {loa_lower:.2f} (cm/s)', color='black', va='top',  fontsize = 25)
plt.text(20, md + 0.2, f'{md:.2f} $\pm$ {sd:.2f} (cm/s)', color='black', va='bottom',  fontsize = 25)

plt.axhline(md, color='red', linestyle='-')
plt.axhline(loa_upper, color='red', linestyle='--')
plt.axhline(loa_lower, color='red', linestyle='--')

plt.ylabel('$Velocity_{TAFE} - Velocity_{Bolus}$', fontsize=30)
plt.xlabel('$Velocity_{Average}$ (cm/s)', fontsize=30)
plt.xticks(size=20)
plt.yticks(size=20)
plt.ylim((-100, 100))
plt.show()
```
